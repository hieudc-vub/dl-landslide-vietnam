# Optional CLI notebook

Use the Windows README workflow first. This notebook invokes the same package and starts real training only when its run cell is executed. Select the project environment; install ipykernel separately if using a notebook host. No author outputs or checkpoints are included.


```python
from pathlib import Path
import subprocess
import sys

REPO = Path.cwd()
if REPO.name == "notebooks":
    REPO = REPO.parent
assert (REPO / "configs/models").is_dir(), "Open from repository root or notebooks/"
DATA_ROOT = REPO / "data/SAMPLE_DATASET_256x256"
CONFIG = REPO / "configs/models/04_unetplusplus_resnet50.json"
OUT = REPO / "runs/04_unetplusplus_resnet50"
DEVICE = "cuda"
```

## Train, calibrate on VAL, evaluate TEST

The output directory must be new. First training use downloads ImageNet weights. CPU execution is available by setting DEVICE to `cpu`, but uses FP32.


```python
subprocess.run([sys.executable, "-m", "landslide", "run",
    "--config", str(CONFIG), "--data-root", str(DATA_ROOT),
    "--out", str(OUT), "--device", DEVICE], cwd=REPO, check=True)
```
